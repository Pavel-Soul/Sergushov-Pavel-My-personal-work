```python
import numpy as np

# --- ДАННЫЕ ИЗ ВАРИАНТА 2 ---

# Структурная матрица торговли A
A = np.array([
    [0.60, 0.20, 0.00],
    [0.40, 0.40, 0.60],
    [0.00, 0.40, 0.40]
])

# Начальный вектор денежных ресурсов x(0)
# В задании он не указан, поэтому выберем произвольный для примера.
# Например, пусть у каждой страны будет по 100 денежных единиц.
x0 = np.array([100.0, 100.0, 100.0])

# --- ВЫЧИСЛЕНИЯ ---

print("--- Численное моделирование торговых циклов ---")
print(f"Матрица торговли A:\n{A}\n")
print(f"Начальный вектор ресурсов x(0): {x0}\n")
print("--- Последовательность векторов x(k) = A * x(k-1) ---")

# Сохраняем текущий вектор, начиная с x(0)
x_current = x0

# Выводим начальное состояние
print(f"x(0) = {[round(val, 4) for val in x_current]}")

# Запускаем цикл для вычисления x(1), x(2), ..., x(12)
for k in range(1, 13):
    # Выполняем основную операцию модели: x(k) = A * x(k-1)
    # Оператор @ в NumPy означает матричное умножение
    x_current = A @ x_current
    
    # Выводим результат на текущем шаге (k), округляя для наглядности
    print(f"x({k}) = {[round(val, 4) for val in x_current]}")

# --- АНАЛИЗ РЕЗУЛЬТАТОВ ---
print("\n--- Анализ сходимости ---")
# Возьмем последний вычисленный вектор x(12)
x_final = x_current
# Найдем соотношение его компонент
ratio_s2_s1 = x_final[1] / x_final[0]
ratio_s3_s1 = x_final[2] / x_final[0]

print(f"Вектор на последнем шаге x(12) ~ {[round(val, 2) for val in x_final]}")
print(f"Соотношение S2/S1 = {round(ratio_s2_s1, 4)}")
print(f"Соотношение S3/S1 = {round(ratio_s3_s1, 4)}")

print("\nСравним с аналитическим равновесным вектором x_p = (3, 6, 4):")
print(f"Соотношение 6/3 = {6/3}")
print(f"Соотношение 4/3 = {round(4/3, 4)}")
print("Как видно, численные результаты сходятся к аналитически найденному соотношению.")
```

```python
import pandas as pd
from rdflib import Graph, URIRef
from rdflib.namespace import RDF
```

```python
g = Graph()
g.parse("university.n3", format="n3")

print(f"Количество троек в загруженном графе: {len(g)}")
```

```python
query = """
PREFIX e: <http://www.fa.ru/entity/>
PREFIX pd: <http://www.fa.ru/prop/direct/>

SELECT DISTINCT ?person ?name
WHERE {
    ?university pd:IsA e:University ;
                pd:rector ?person .
}
"""

res = g.query(query)

# Преобразуем результат в DataFrame
df = pd.DataFrame(res, columns=["person", "name"])
df["birthDate"] = None
df["name"] = df["person"].apply(lambda x: x.split('/')[-1].replace('_', ' '))
df
```

4\. Напишите и выполните SPARQL запрос, который возвращает количество студентов в каждом университете в различные годы.

- [ ] Проверено на семинаре

```python
query = """
PREFIX p: <http://www.fa.ru/prop/>
PREFIX ps: <http://www.fa.ru/prop/statement/>
PREFIX pq: <http://www.fa.ru/prop/qualifier/>
PREFIX pd: <http://www.fa.ru/prop/direct/>

SELECT ?university ?year ?students
WHERE {
    ?university pd:IsA <http://www.fa.ru/entity/University> ;
                p:studentSize ?sizeStmt .
    ?sizeStmt pq:atYear ?year ;
              ps:studentSize ?students .
}
ORDER BY ?university ?year
"""

res = g.query(query)
df4 = pd.DataFrame(res, columns=["university", "year", "students"])
df4["university_name"] = df4["university"].apply(lambda x: x.split('/')[-1])
df4
```

- [ ] Проверено на семинаре

```python
# Словарь: URI ректора - русское имя
rector_ru_names = {
    "http://www.fa.ru/entity/Viktor_Sadovnichiy": "Виктор Садовничий",
    "http://www.fa.ru/entity/Anatoly_Aleksandrov": "Анатолий Александров",
    "http://www.fa.ru/entity/Nikolay_Kropachev": "Николай Кропачёв",
    "http://www.fa.ru/entity/Andrey_Rudskoy": "Андрей Рудской"
}

# SPARQL-запрос: получаем всех ректоров
query6 = """
PREFIX e: <http://www.fa.ru/entity/>
PREFIX pd: <http://www.fa.ru/prop/direct/>

SELECT ?rector
WHERE {
    ?university pd:IsA e:University ;
                pd:rector ?rector .
}
"""
res6 = g.query(query6)

# Преобразуем в DataFrame
df6 = pd.DataFrame(res6, columns=["rector"])

# Функция для извлечения имени из URI
def extract_name_en(uri):
    return str(uri).split('/')[-1].replace('_', ' ')

# Добавляем колонки
df6["name_en"] = df6["rector"].apply(extract_name_en)
df6["name_ru"] = df6["rector"].apply(lambda x: rector_ru_names.get(str(x)))

# Фильтруем только тех, у кого есть имя на английском
df6 = df6[df6["name_en"].notna()]

df6
```

8\. Напишите и выполните SPARQL запрос, который для каждого университета возвращает три значения: название на русском языке, минимальное количество студентов за все годы, максимальное количество студентов за все годы.

- [ ] Проверено на семинаре

```python
# Словарь: URI университета → русское название
univ_ru_names = {
    "http://www.fa.ru/entity/BaumanMoscowStateTechnicalUniversity": "Московский государственный технический университет имени Баумана",
    "http://www.fa.ru/entity/MoscowStateUniversity": "Московский государственный университет",
    "http://www.fa.ru/entity/SaintPetersburgStateUniversity": "Санкт-Петербургский государственный университет",
    "http://www.fa.ru/entity/SpbPolytechnicUniversity": "Политехнический университет СПб"
}

# SPARQL-запрос: берем университеты и все значения studentSize
query8 = """
PREFIX e: <http://www.fa.ru/entity/>
PREFIX pd: <http://www.fa.ru/prop/direct/>
PREFIX p: <http://www.fa.ru/prop/>
PREFIX ps: <http://www.fa.ru/prop/statement/>

SELECT ?university ?students
WHERE {
    ?university pd:IsA e:University ;
                p:studentSize ?stmt .
    ?stmt ps:studentSize ?students .
}
"""
res8 = g.query(query8)

# Преобразуем в DataFrame
df8 = pd.DataFrame(res8, columns=["university", "students"])

# Конвертируем students в int
df8["students"] = df8["students"].astype(int)

# Группировка по университету
df8_grouped = df8.groupby("university")["students"].agg(minStudents="min", maxStudents="max").reset_index()

# Добавляем русские названия
df8_grouped["name_ru"] = df8_grouped["university"].apply(lambda x: univ_ru_names.get(str(x)))

# Переупорядочим колонки
df8_grouped = df8_grouped[["name_ru", "minStudents", "maxStudents"]]

df8_grouped
```
