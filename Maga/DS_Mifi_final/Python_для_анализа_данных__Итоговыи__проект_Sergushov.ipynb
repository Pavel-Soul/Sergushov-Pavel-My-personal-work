3. Следующим этапом преобразуем признак **"Опыт работы"**. Его текущий формат - это: **<Опыт работы: n лет m месяцев, периоды работы в различных компаниях…>**. 

Из столбца нам необходимо выделить общий опыт работы соискателя в месяцах, новый признак назовем "Опыт работы (месяц)"

Для начала обсудим условия решения задачи:
* Во-первых, в данном признаке есть пропуски. Условимся, что если мы встречаем пропуск, оставляем его как есть (функция-преобразование возвращает NaN)
* Во-вторых, в данном признаке есть скрытые пропуски. Для некоторых соискателей в столбце стоит значения "Не указано". Их тоже обозначим как NaN (функция-преобразование возвращает NaN)
* В-третьих, нас не интересует информация, которая описывается после указания опыта работы (периоды работы в различных компаниях)
* В-четвертых, у нас есть проблема: опыт работы может быть представлен только в годах или только месяцах. Например, можно встретить следующие варианты:
    * Опыт работы 3 года 2 месяца…

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы и удалите столбец **"Опыт работы"** из таблицы.


```python
def get_experience(text):
    if pd.isna(text) or text.strip() == "Не указано":
        return np.nan

    words = text.split(maxsplit=6)
    number = int(words[2])

    if words[3].startswith("месяц"):
        return number

    months = number * 12
    if len(words) >= 6 and words[4].isdigit() and words[5].startswith("месяц"):
        months += int(words[4])

    return months
```

```python
data["Опыт работы (месяц)"] = data["Опыт работы"].apply(get_experience)

data = data.drop(columns=["Опыт работы"])

print("Медиана:", data["Опыт работы (месяц)"].median())
print("Количество пропусков:", data["Опыт работы (месяц)"].isna().sum())
```

Итак, давайте обсудим возможный алгоритм преобразования: 
1. Перевести признак "Обновление резюме" из таблицы с резюме в формат datetime и достать из него дату. В тот же формат привести признак "date" из таблицы с валютами.
2. Выделить из столбца "ЗП" сумму желаемой заработной платы и наименование валюты, в которой она исчисляется. Наименование валюты перевести в стандарт ISO согласно с таблицей выше.
3. Присоединить к таблице с резюме таблицу с курсами по столбцам с датой и названием валюты (подумайте, какой тип объединения надо выбрать, чтобы в таблице с резюме сохранились данные о заработной плате, изначально представленной в рублях). Значение close для рубля заполнить единицей 1 (курс рубля самого к себе)
4. Умножить сумму желаемой заработной платы на присоединенный курс валюты (close) и разделить на пропорцию (обратите внимание на пропуски после объединения в этих столбцах), результат занести в новый столбец "ЗП (руб)".


```python
rates = pd.read_csv(
    "/Users/pavelsergusov/Desktop/git_pavel/Maga/DS_Mifi_final/ExchangeRates.csv",
    usecols=["currency", "date", "close", "proportion"])
rates.head()
```

```python
rates["date"] = pd.to_datetime(rates["date"], format="%d/%m/%y")
data["Обновление резюме"] = pd.to_datetime(data["Обновление резюме"], format="%d.%m.%Y %H:%M").dt.normalize()

# Разделяем зарплату на сумму и валюту
salary = data["ЗП"].str.split(expand=True)
data["amount"] = salary[0].astype(float)
data["currency"] = salary[1].replace({
    "руб.": "RUB", "бел.руб.": "BYN", "грн.": "UAH", "сум": "UZS"
})

# Для каждого резюме находим курс по валюте и дате
data = data.merge(
    rates,
    how="left",
    left_on=["currency", "Обновление резюме"],
    right_on=["currency", "date"],
    validate="many_to_one"
)

# Для зарплат в рублях курс и номинал равны 1
data.loc[data["currency"] == "RUB", ["close", "proportion"]] = 1
data["ЗП (руб)"] = data["amount"] * data["close"] / data["proportion"]

print("Медианная зарплата в рублях:", data["ЗП (руб)"].median())
print("Количество пропусков:", data["ЗП (руб)"].isna().sum())

# Удаляем исходную зарплату и промежуточные столбцы
data = data.drop(columns=["ЗП", "amount", "currency", "date", "close", "proportion"])
```

```python
def plot_distribution(values, label, step):
    values = values.dropna()
    fig, ax = plt.subplots(1, 2, figsize=(11, 3.2), sharex=True)

    sns.histplot(x=values, binwidth=step, ax=ax[0])
    ax[0].set(xlabel=label, ylabel="Число резюме", title="Гистограмма")

    sns.boxplot(x=values, ax=ax[1], fliersize=3)
    ax[1].set(xlabel=label, title="Коробчатая диаграмма")

    plt.tight_layout()
    plt.show()


age = data["Возраст"]
plot_distribution(age, "Возраст, лет", 1)

print("Минимум, мода и максимум:", age.min(), age.mode().iat[0], age.max())
print(f"От 25 до 45 лет: {age.between(25, 45).mean():.1%}")
```

В какую сторону асимметрично логарифмическое распределение? Напишите об этом в комментарии к графику.
Найдите выбросы с помощью метода z-отклонения и удалите их из данных, используйте логарифмический масштаб. Давайте сделаем послабление на **1 сигму** (возьмите 4 сигмы) в **правую сторону**.

Выведите таблицу с полученными выбросами и оцените, с каким возрастом соискатели попадают под категорию выбросов?

```python
log_age = np.log(data["Возраст"])
mean = log_age.mean()
std = log_age.std()

lower = mean - 3 * std
upper = mean + 4 * std

fig, ax = plt.subplots(figsize=(9, 4))
sns.histplot(x=log_age, bins=40, ax=ax)

ax.axvline(mean, color="black", label="Среднее")
ax.axvline(lower, color="red", linestyle="--", label="−3σ")
ax.axvline(mean + 3 * std, color="gray", linestyle=":", label="+3σ")
ax.axvline(upper, color="red", linestyle="--", label="+4σ")

ax.set(xlabel="Натуральный логарифм возраста", ylabel="Число резюме",
       title="Распределение возраста в логарифмическом масштабе")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
age_outliers = ~log_age.between(lower, upper)
display(data.loc[age_outliers, ["Возраст", "Опыт работы (месяц)", "ЗП (руб)"]]
        .sort_values("Возраст"))

print("Асимметрия:", round(log_age.skew(), 2))
print("Границы в годах:", round(np.exp(lower), 2), round(np.exp(upper), 2))
print("Удалено резюме:", age_outliers.sum())

data = data.loc[~age_outliers].copy()
print("Итоговый размер таблицы:", data.shape)
```

После логарифмирования распределение возраста немного скошено вправо. Это видно по более длинному правому хвосту и положительному коэффициенту асимметрии 0,45. При границах от −3σ до +4σ допустимый интервал составляет примерно от 15,66 до 78,88 года. За его пределы попали три резюме: два соискателя указали возраст 15 лет, один — 100 лет.
