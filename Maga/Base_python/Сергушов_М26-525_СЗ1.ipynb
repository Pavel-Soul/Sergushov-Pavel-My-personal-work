### Задание 2
Сжатие строки

```python
def compress_string(text):
    if text == "":
        return ""

    result = ""
    count = 1

    for i in range(1, len(text)):
        if text[i] == text[i - 1]:
            count += 1
        else:
            result += text[i - 1]
            if count > 1:
                result += str(count)
            count = 1

    # Добавляем последнюю серию символов.
    result += text[-1]
    if count > 1:
        result += str(count)

    if len(result) <= len(text):
        return result
    return text
```

```python
exmpl = ["aaabbc", "abcd", "", "a", "aaaaaaaaaaaa", "   !!!"]

for text in exmpl:
    print((text), "→", (compress_string(text)))
```

### Задание 3
Проверка скобок

```python
def check_skob(text):
    stack = []
    pairs = {")": "(", "]": "[", "}": "{"}

    for a in text:
        if a in "([{":
            stack.append(a)

        elif a in ")]}":
            if not stack:
                return False

            if stack.pop() != pairs[a]:
                return False

    return len(stack) == 0
```

```python
exmpl2 = ["()", "([]){}", "([)]", "(()", ")(", "", "a + (b * c)"]

for text in exmpl2:
    print((text), "→", check_skob(text))
```

### Задание 4
Генератор паролей


```python
import random
import string
```

```python
def generate_password(length):

    letters = string.ascii_letters
    digits = string.digits
    symbols = "!@#$%^&*"
    all_chars = letters + digits + symbols

    # Сразу добавляем по одному символу каждого типа.
    password = [
        random.choice(letters),
        random.choice(digits),
        random.choice(symbols)
    ]

    # Заполняем оставшиеся места.
    for _ in range(length - 3):
        password.append(random.choice(all_chars))

    random.shuffle(password)

    return "".join(password)
```

```python
length = int(input("Введите длину пароля: "))
password = generate_password(length)

print("Пароль:", password)
print("Длина:", len(password))
```

### Задание 5
Почти палиндром

```python
def palindrome(text):
    text = "".join(text.lower().split())

    if text == text[::-1]:
        return True

    # По очереди пробуем удалить одну букву.
    for i in range(len(text)):
        shortened = text[:i] + text[i + 1:]

        if shortened == shortened[::-1]:
            return True

    return False
```

```python
exmpl3 = ["казак","lol", "поп", "А роза упала на лапу Азора", "abca", "abc", "", "а"]

for text in exmpl3:
    print((text), "→", palindrome(text))
```

### Задание 6. 
Задача Шредингера

```python
import re
```

```python
def erase_text(file_path, percent):
    with open(file_path, encoding="utf-8") as file:
        text = file.read()

    words = list(re.finditer(r"\b\w+\b", text))

    count = round(len(words) * percent / 100)
    selected = random.sample(words, count)

    chars = list(text)

    for word in selected:
        for i in range(word.start(), word.end()):
            chars[i] = " "

    return "".join(chars), count
```
