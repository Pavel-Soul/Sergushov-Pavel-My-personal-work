## Задание 1. Найти производные функций

### а) $y=e^{x\ln(3x+1)}$

Сначала продифференцируем показатель степени. В нём переменная $x$ умножается на логарифм, поэтому используем правило произведения:

$$
\bigl(x\ln(3x+1)\bigr)'
= \ln(3x+1) + x\cdot\frac{3}{3x+1}.
$$

Дробь $\frac{3}{3x+1}$ получается при дифференцировании логарифма. Тройка в числителе — это производная выражения $3x+1$.

Теперь умножаем исходную экспоненту на производную её показателя:

$$
\boxed{y'=e^{x\ln(3x+1)}
\left(\ln(3x+1)+\frac{3x}{3x+1}\right)}.
$$

Логарифм определён при $3x+1>0$, то есть при $x>-\frac13$.

### б) $y=e^{x^2}\sqrt{x^3+4x^2-7}$

Здесь перемножаются экспонента и корень. Найдём производную каждого множителя:

$$
\left(e^{x^2}\right)'=2xe^{x^2},
$$

$$
\left(\sqrt{x^3+4x^2-7}\right)'
=\frac{3x^2+8x}{2\sqrt{x^3+4x^2-7}}.
$$

По правилу произведения сначала дифференцируем экспоненту и оставляем корень, затем дифференцируем корень и оставляем экспоненту. Складываем эти два слагаемых и выносим $e^{x^2}$ за скобки:

$$
\boxed{
y'=e^{x^2}\left(
2x\sqrt{x^3+4x^2-7}
+\frac{3x^2+8x}{2\sqrt{x^3+4x^2-7}}
\right)
}.
$$

Эта формула для производной применима при $x^3+4x^2-7>0$, поскольку корень стоит в знаменателе.

### в) $y=\dfrac{x\sqrt{1-x^2}}{1+x^2}$

Обозначим числитель через $u=x\sqrt{1-x^2}$, а знаменатель через $v=1+x^2$. В числителе снова применяем правило произведения:

$$
u'=\sqrt{1-x^2}-\frac{x^2}{\sqrt{1-x^2}}
=\frac{1-2x^2}{\sqrt{1-x^2}}.
$$

У знаменателя производная $v'=2x$. Подставляем обе производные в формулу для производной частного:

$$
y'=\frac{u'v-uv'}{v^2}
=\frac{(1-2x^2)(1+x^2)-2x^2(1-x^2)}
{(1+x^2)^2\sqrt{1-x^2}}.
$$

Раскрываем скобки в числителе. Слагаемые с $x^4$ сокращаются, и остаётся $1-3x^2$.

$$
\boxed{y'=\frac{1-3x^2}{(1+x^2)^2\sqrt{1-x^2}}},\qquad -1<x<1.
$$

## Задание 2. Безусловные экстремумы

Найдём экстремумы функции

$$
u(x,y,z)=3x^3+y^2+z^2+6xy-2z+1.
$$

### 1. Найдём стационарные точки

Вычислим частные производные и приравняем их к нулю.

$$
\begin{cases}
u_x=9x^2+6y=0,\\
u_y=2y+6x=0,\\
u_z=2z-2=0.
\end{cases}
$$

Из второго уравнения получаем $y=-3x$, а из третьего — $z=1$. Подставим $y=-3x$ в первое уравнение.

$$
9x^2-18x=0
\quad\Longrightarrow\quad
9x(x-2)=0.
$$

Получаем $x=0$ или $x=2$. Подставив эти значения в $y=-3x$, найдём две стационарные точки.

$$
A=(0,0,1),\qquad B=(2,-6,1).
$$

### 2. Определим тип каждой точки

Составим матрицу Гессе из вторых частных производных.

$$
H(x,y,z)=
\begin{pmatrix}
18x & 6 & 0\\
6 & 2 & 0\\
0 & 0 & 2
\end{pmatrix}.
$$

**Точка $A=(0,0,1)$.** Подставим $x=0$. Определитель верхнего левого блока размером $2\times 2$ равен

$$
\Delta_2=
\begin{vmatrix}
0 & 6\\
6 & 2
\end{vmatrix}
=-36.
$$

Отрицательный определитель этого блока означает, что при малых отклонениях в плоскости $z=1$ функция в одних направлениях растёт, а в других убывает. Значит, $A$ — седловая точка.

**Точка $B=(2,-6,1)$.** При $x=2$ угловые миноры матрицы Гессе равны

$$
\Delta_1=36,\qquad
\Delta_2=36\cdot 2-6^2=36,\qquad
\Delta_3=2\cdot 36=72.
$$

Все три минора положительны. По критерию Сильвестра матрица Гессе положительно определена, поэтому в точке $B$ функция имеет строгий локальный минимум. Его значение равно

$$
u(2,-6,1)=24+36+1-72-2+1=-12.
$$

**Ответ.** Строгий локальный минимум достигается в точке $(2,-6,1)$ и равен $-12$. Точка $(0,0,1)$ — седловая. Локальных максимумов нет.

Глобальных экстремумов тоже нет. При $y=0$ и $z=1$ функция принимает вид $u(x,0,1)=3x^3$ и не ограничена ни сверху, ни снизу.

## Задание 3. Условные экстремумы

Найдём минимум и максимум функции $z=e^{-x-y}$ на эллипсе

$$
\frac{x^2}{4}+y^2=5.
$$

### 1. Найдём точки методом Лагранжа

Запишем функцию Лагранжа.

$$
L(x,y,\lambda)=e^{-x-y}
+\lambda\left(\frac{x^2}{4}+y^2-5\right).
$$

Приравняем её частные производные к нулю.

$$
\begin{cases}
-e^{-x-y}+\lambda\dfrac{x}{2}=0,\\[4pt]
-e^{-x-y}+2\lambda y=0,\\[4pt]
\dfrac{x^2}{4}+y^2-5=0.
\end{cases}
$$

Из первых двух уравнений получаем $\lambda x/2=2\lambda y$. Экспонента всегда положительна, поэтому $\lambda\ne 0$. Сокращаем на $\lambda$ и получаем $x=4y$.

Подставим $x=4y$ в уравнение эллипса.

$$
\frac{(4y)^2}{4}+y^2=5
\quad\Longrightarrow\quad
5y^2=5.
$$

При $y=1$ получаем $x=4$, а при $y=-1$ получаем $x=-4$. Из второго уравнения системы также найдём значения $\lambda$ в обеих точках.

$$
A=(4,1),\qquad \lambda_A=\frac{e^{-5}}{2},
$$

$$
B=(-4,-1),\qquad \lambda_B=-\frac{e^5}{2}.
$$

### 2. Проверим точки с помощью матрицы Гессе

Матрица Гессе функции Лагранжа по переменным $x$ и $y$ имеет вид

$$
H_L=
\begin{pmatrix}
e^{-x-y}+\dfrac{\lambda}{2} & e^{-x-y}\\[4pt]
e^{-x-y} & e^{-x-y}+2\lambda
\end{pmatrix}.
$$

Двигаться можно только вдоль эллипса, поэтому проверим знак квадратичной формы в направлении касательной. Градиент ограничения равен $(x/2,\,2y)$, значит, для касательного вектора $(h,k)$ выполняется

$$
\frac{x}{2}h+2yk=0.
$$

В обеих точках это условие сводится к $h+k=0$. Возьмём касательный вектор $(t,-t)$, где $t\ne 0$. Тогда квадратичная форма равна

$$
\begin{aligned}
Q(t,-t)
&=\left(e^{-x-y}+\frac{\lambda}{2}\right)t^2
-2e^{-x-y}t^2
+\left(e^{-x-y}+2\lambda\right)t^2\\
&=\frac{5}{2}\lambda t^2.
\end{aligned}
$$

В точке $A$ множитель $\lambda_A>0$, поэтому $Q>0$ и здесь достигается условный минимум. В точке $B$ множитель $\lambda_B<0$, поэтому $Q<0$ и здесь достигается условный максимум.

Эллипс замкнут и ограничен, а функция непрерывна. Значит, на эллипсе достигаются наименьшее и наибольшее значения. Мы нашли обе точки, в которых они могут достигаться.

**Ответ.** Минимум $z_{\min}=e^{-5}$ достигается в точке $(4,1)$. Максимум $z_{\max}=e^5$ достигается в точке $(-4,-1)$.

```python
import pandas as pd
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_csv(
    "asset-v1_SkillFactory+MIFIML-1sem+2024+type@asset+block@sf_exam.csv"
)

features = ["Age_Group", "Test"]
target = "Status"

train, test = train_test_split(
    data, test_size=0.25, random_state=42, stratify=data[target]
)

print(f"Обучающая выборка: {len(train)}")
print(f"Тестовая выборка: {len(test)}")
```

```python
prior = train[target].value_counts(normalize=True)
probabilities = {}

for feature in features:
    counts = pd.crosstab(train[feature], train[target])

    # Сглаживание Лапласа, чтобы избежать нулевых вероятностей
    counts = counts + 1
    probabilities[feature] = counts / counts.sum(axis=0)


def predict(row):
    scores = prior.copy()

    for feature in features:
        scores *= probabilities[feature].loc[row[feature]]

    return scores.idxmax()


y_true = test[target]
y_pred = test[features].apply(predict, axis=1)
```

```python
labels = ["Not_infected", "Infected"]

print(f"Accuracy: {accuracy_score(y_true, y_pred):.2%}")
print(classification_report(y_true, y_pred, labels=labels, digits=3))

matrix = pd.DataFrame(
    confusion_matrix(y_true, y_pred, labels=labels),
    index=pd.Index(labels, name="Факт"),
    columns=pd.Index(labels, name="Прогноз"),
)
display(matrix)
```

На тестовой выборке модель дала 54 правильных ответа из 70. Доля верных ответов составила **77,14%**.

Модель распознала 25 из 36 записей класса `Infected`. Полнота для этого класса составила **69,4%**. Из 16 ошибок 11 пришлись на случаи, где модель пропустила `Infected`.

## Задание 5. Метод Ньютона

Дана функция

$$
f(x)=x^4+3x^3-12x^2+7x-2.
$$

В точке экстремума первая производная равна нулю. Поэтому методом Ньютона решаем уравнение $f'(x)=0$. Для этого понадобятся обе производные.

$$
f'(x)=4x^3+9x^2-24x+7,
$$

$$
f''(x)=12x^2+18x-24.
$$

На каждом шаге вычисляем новое приближение по формуле

$$
x_{k+1}=x_k-\frac{f'(x_k)}{f''(x_k)}.
$$

Начинаем с $x_0=10$. Останавливаемся, когда разница между двумя соседними приближениями становится меньше $0{,}0001$.

```python
def f(x):
    return x**4 + 3*x**3 - 12*x**2 + 7*x - 2


def df(x):
    return 4*x**3 + 9*x**2 - 24*x + 7


def d2f(x):
    return 12*x**2 + 18*x - 24


x = 10.0
eps = 0.0001
iterations = 0

while True:
    x_next = x - df(x) / d2f(x)
    step = abs(x_next - x)
    x = x_next
    iterations += 1

    if step < eps:
        break

print(f"Итераций: {iterations}")
print(f"x = {x:.6f}")
print(f"f(x) = {f(x):.6f}")
print(f"f'(x) = {df(x):.3e}")
print(f"f''(x) = {d2f(x):.6f}")
print(f"Последний шаг: {step:.3e}")
```

### Вывод

Метод сошёлся за 9 итераций. Последний шаг составил примерно $0{,}00001274$, что меньше заданной точности $0{,}0001$.

В найденной точке $x\approx 1{,}310136$ первая производная близка к нулю, а вторая положительна: $f''(x)\approx 20{,}179904$. Значит, это точка локального минимума. Значение функции в ней равно примерно $-3{,}733928$.

## Задание 6. Распределение числа спам-писем

Пусть $X$ — количество спам-писем, полученных за день.

### 1. Вероятность получить 7 писем

Сумма вероятностей всех возможных значений $X$ равна единице. Поэтому из единицы вычитаем сумму известных вероятностей.

$$
P(X=7)=1-(0{,}1+0{,}3+0{,}1+0{,}25+0{,}05+0{,}05+0{,}1)
=0{,}05.
$$

Значит, вероятность получить ровно 7 писем составляет **5%**.

### 2. Математическое ожидание

Умножаем каждое количество писем на его вероятность и складываем результаты.

$$
\begin{aligned}
E(X)
&=0\cdot 0{,}1+1\cdot 0{,}3+2\cdot 0{,}1+3\cdot 0{,}25\\
&\quad+4\cdot 0{,}05+5\cdot 0{,}05+6\cdot 0{,}1+7\cdot 0{,}05\\
&=2{,}65.
\end{aligned}
$$

В среднем за день приходит **2,65 спам-письма**.

### 3. Дисперсия

Воспользуемся формулой $D(X)=E(X^2)-[E(X)]^2$. Чтобы найти $E(X^2)$, умножаем квадрат каждого значения на его вероятность.

$$
\begin{aligned}
E(X^2)
&=0^2\cdot 0{,}1+1^2\cdot 0{,}3+2^2\cdot 0{,}1+3^2\cdot 0{,}25\\
&\quad+4^2\cdot 0{,}05+5^2\cdot 0{,}05+6^2\cdot 0{,}1+7^2\cdot 0{,}05\\
&=11{,}05.
\end{aligned}
$$

Подставляем значения и округляем результат до сотых.

$$
D(X)=11{,}05-(2{,}65)^2=4{,}0275\approx 4{,}03.
$$

**Ответ.** $P(X=7)=0{,}05$, $E(X)=2{,}65$, $D(X)\approx 4{,}03$.

## Задание 7. Вероятность поступления звонков

За 8 часов в колл-центр поступает в среднем 16 звонков. Будем считать, что звонки образуют пуассоновский поток с постоянной интенсивностью. Среднее число звонков за час равно

$$
\lambda=\frac{16}{8}=2.
$$

Обозначим число звонков за час через $X$. По условию звонков должно быть не меньше двух и не больше четырёх. Значит, подходят три случая: $X=2$, $X=3$ и $X=4$.

По формуле Пуассона

$$
P(X=k)=\frac{\lambda^k e^{-\lambda}}{k!}.
$$

Сложим вероятности трёх подходящих случаев.

$$
\begin{aligned}
P(2\le X\le 4)
&=P(X=2)+P(X=3)+P(X=4)\\
&=e^{-2}\left(\frac{2^2}{2!}+\frac{2^3}{3!}+\frac{2^4}{4!}\right)\\
&=e^{-2}\left(2+\frac{4}{3}+\frac{2}{3}\right)\\
&=4e^{-2}\approx 0{,}541341.
\end{aligned}
$$

### Расчёт на Python

Вычислим $e^\lambda$ по ряду $1+\lambda+\frac{\lambda^2}{2!}+\ldots$. Начнём с единицы и получим каждый следующий член, умножая предыдущий на $\lambda/n$.

Вероятность нуля звонков равна $P(X=0)=1/e^\lambda$. Чтобы получить вероятность ровно $k$ звонков, умножим предыдущую вероятность на $\lambda/k$. В итоговую сумму добавим только значения для $k=2,3,4$.

```python
rate = 16 / 8

# Вычисляем e^rate по первым 21 членам ряда
term = 1.0
exp_rate = term
n = 1

while n <= 20:
    term *= rate / n
    exp_rate += term
    n += 1

# Начинаем с P(X = 0) и последовательно вычисляем P(X = k)
p = 1 / exp_rate
probability = 0.0
k = 1

while k <= 4:
    p *= rate / k

    if k >= 2:
        probability += p

    k += 1

print(f"Вероятность: {probability:.9f}")
print(f"Ответ до тысячных: {probability:.3f}")
```
