### Задание 1. Ассоциативность умножения матриц


Вычислим выражение $X^T(XY)$ двумя способами и сравним количество операций.

#### Исходные данные

Берём первые три квартиры и признаки «Комнаты», «Площадь», «Этаж»:

$$
X =
\begin{pmatrix}
3 & 51 & 3 \\
1 & 30 & 1 \\
2 & 45 & 2
\end{pmatrix},
\qquad
Y =
\begin{pmatrix}
2200 \\
1600 \\
1900
\end{pmatrix}.
$$



Транспонированная матрица:

$$
X^T =
\begin{pmatrix}
3 & 1 & 2 \\
51 & 30 & 45 \\
3 & 1 & 2
\end{pmatrix}.
$$

#### Способ 1. Сначала вычисляем XY

$$
u = XY =
\begin{pmatrix}
3 \cdot 2200 + 51 \cdot 1600 + 3 \cdot 1900 \\
1 \cdot 2200 + 30 \cdot 1600 + 1 \cdot 1900 \\
2 \cdot 2200 + 45 \cdot 1600 + 2 \cdot 1900
\end{pmatrix}
=
\begin{pmatrix}
93900 \\
52100 \\
80200
\end{pmatrix}.
$$

Размер $u$ — $3 \times 1$.

$$
X^T u =
\begin{pmatrix}
3 \cdot 93900 + 1 \cdot 52100 + 2 \cdot 80200 \\
51 \cdot 93900 + 30 \cdot 52100 + 45 \cdot 80200 \\
3 \cdot 93900 + 1 \cdot 52100 + 2 \cdot 80200
\end{pmatrix}
=
\begin{pmatrix}
494200 \\
9960900 \\
494200
\end{pmatrix}.
$$

Размер результата — $3 \times 1$.

#### Способ 2. Сначала вычисляем XᵀX

$$
M = X^T X =
\begin{pmatrix}
14 & 273 & 14 \\
273 & 5526 & 273 \\
14 & 273 & 14
\end{pmatrix}.
$$

Размер $M$ — $3 \times 3$.

Примеры вычисления элементов:

$$
M_{11} = 3 \cdot 3 + 1 \cdot 1 + 2 \cdot 2 = 14.
$$

$$
M_{12} = 3 \cdot 51 + 1 \cdot 30 + 2 \cdot 45 = 273.
$$

$$
M_{22} = 51 \cdot 51 + 30 \cdot 30 + 45 \cdot 45 = 5526.
$$

Умножаем полученную матрицу на вектор цен:

$$
MY =
\begin{pmatrix}
14 \cdot 2200 + 273 \cdot 1600 + 14 \cdot 1900 \\
273 \cdot 2200 + 5526 \cdot 1600 + 273 \cdot 1900 \\
14 \cdot 2200 + 273 \cdot 1600 + 14 \cdot 1900
\end{pmatrix}
=
\begin{pmatrix}
494200 \\
9960900 \\
494200
\end{pmatrix}.
$$

Размер результата — $3 \times 1$. Результаты обоих способов совпали:

$$
X^T(XY) = (X^T X)Y.
$$

#### Сравнение количества операций

При обычном умножении каждый элемент результата требует
3 скалярных умножения и 2 сложения. Сокращения за счёт
повторяющихся значений и симметрии не учитываем.

| Способ | Этап | Умножения | Сложения |
|---|---|---:|---:|
| 1 | Вычисление $XY$: 3 элемента | 9 | 6 |
| 1 | Вычисление $X^T u$: 3 элемента | 9 | 6 |
| **Итого, способ 1** | | **18** | **12** |
| 2 | Вычисление $X^T X$: 9 элементов | 27 | 18 |
| 2 | Вычисление $MY$: 3 элемента | 9 | 6 |
| **Итого, способ 2** | | **36** | **24** |

#### Вывод

Оба порядка группировки дают одинаковый результат, но первый способ
требует вдвое меньше скалярных операций: 30 вместо 60.
Для больших матриц выгодный порядок умножения позволяет уменьшить
время вычислений и объём промежуточных данных.

### Задание 4. Обратная матрица

Составим матрицу X из первых трёх квартир.
Столбцы соответствуют признакам «Комнаты» и «Площадь».

$$
X =
\begin{pmatrix}
3 & 51 \\
1 & 30 \\
2 & 45
\end{pmatrix},
\qquad
X^T =
\begin{pmatrix}
3 & 1 & 2 \\
51 & 30 & 45
\end{pmatrix}.
$$


#### 1. Вычисление матрицы A

$$
A = X^T X =
\begin{pmatrix}
3 \cdot 3 + 1 \cdot 1 + 2 \cdot 2
&
3 \cdot 51 + 1 \cdot 30 + 2 \cdot 45
\\
51 \cdot 3 + 30 \cdot 1 + 45 \cdot 2
&
51 \cdot 51 + 30 \cdot 30 + 45 \cdot 45
\end{pmatrix}
=
\begin{pmatrix}
14 & 273 \\
273 & 5526
\end{pmatrix}.
$$

Размер $A$ — $2 \times 2$.

#### 2. Определитель

Для матрицы размера $2 \times 2$:

$$
\det(A) = ad - bc.
$$

Подставляем значения:

$$
\det(A) = 14 \cdot 5526 - 273 \cdot 273
= 77364 - 74529 = 2835.
$$

Определитель не равен нулю, поэтому обратная матрица существует.

#### 3. Обратная матрица

Используем формулу:

$$
A^{-1} =
\frac{1}{ad-bc}
\begin{pmatrix}
d & -b \\
-c & a
\end{pmatrix}.
$$

Получаем:

$$
A^{-1} =
\frac{1}{2835}
\begin{pmatrix}
5526 & -273 \\
-273 & 14
\end{pmatrix}.
$$

Размер $A^{-1}$ — $2 \times 2$.

#### 4. Если бы определитель был равен нулю

При $\det(A)=0$ столбцы матрицы X были бы линейно зависимыми.
Обратная матрица $(X^T X)^{-1}$ не существовала бы,
поэтому вычислить коэффициенты по формуле
$\hat{\beta}=(X^T X)^{-1}X^T Y$ было бы невозможно.

#### Вывод

Определитель матрицы A равен 2835, поэтому она обратима.
Для выбранных признаков можно использовать формулу
коэффициентов регрессии с обратной матрицей.

### Задание 5. Построение модели методом OLS

Используем все пять квартир и два признака: «Площадь» и «Центр».
Добавляем первый столбец единиц для расчёта свободного члена.

Коэффициенты вычисляем методом наименьших квадратов:

$$
\hat{\beta}=(X^T X)^{-1}X^T Y.
$$

```python
# Столбцы X: единицы, площадь, центр.
X = np.column_stack([
    np.ones(len(data)),
    data["Площадь"],
    data["Центр"]
])

Y = data[["Цена"]].to_numpy()

print("Матрица X:")
print(X)
print("Размер X:", X.shape)

print("\nВектор Y:")
print(Y)
print("Размер Y:", Y.shape)

A = X.T @ X
print("\nA = XᵀX:")
print(A)
print("Размер A:", A.shape)

det_A = np.linalg.det(A)
print("\nОпределитель A:", round(det_A, 3))

A_inverse = np.linalg.inv(A)
print("\nОбратная матрица A:")
print(A_inverse.round(6))

XTY = X.T @ Y
print("\nXᵀY:")
print(XTY)
print("Размер XᵀY:", XTY.shape)

beta = A_inverse @ XTY
print("\nКоэффициенты модели:")
print("β₀ =", round(beta[0, 0], 3))
print("β₁ =", round(beta[1, 0], 3))
print("β₂ =", round(beta[2, 0], 3))
```
