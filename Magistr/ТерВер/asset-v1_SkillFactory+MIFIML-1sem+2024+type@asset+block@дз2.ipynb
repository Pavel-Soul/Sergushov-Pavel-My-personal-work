**Задача 1**. (11 баллов)

В известном магазине техники представлено очень много экземпляров различной бытовой техники двух видов: с возможностью подключения к системе "Умный дом" и без такой возможности. Причем, известно, что эти два вида техники представлены в пропорции 1 : 3 соответственно. Случайным образом выбрали 7 экземпляров техники для розыгрыша среди участниц конкурса "Мисс Рукоделье - 2024".  Какова вероятность, что среди них не более двух экземпляров с возможностью подключения к системе "Умный дом"?

Подробно распишите решение и обоснуйте выбор распределения для моделирования сюжета задачи. Ответ округлите до трёх знаков.


**Решение.** Пусть $X$ — число выбранных устройств с подключением к системе «Умный дом». Из пропорции $1:3$ следует, что доля таких устройств равна $p=1/(1+3)=1/4$. Доля остальных равна $3/4$.

Для каждого из семи устройств возможны два исхода. При выборе без возвращения доля подходящей техники меняется, но по условию запас очень велик по сравнению с выборкой из семи устройств. Поэтому считаем выборы приближённо независимыми и используем биномиальное распределение с параметрами $n=7$ и $p=1/4$.

«Не более двух» означает ноль, одно или два устройства. Эти исходы несовместимы, поэтому складываем их вероятности. Для ровно $k$ подходящих устройств число способов выбрать их места среди семи равно $\binom{7}{k}$.

$$P(X\le 2)=(3/4)^7+7(1/4)(3/4)^6+21(1/4)^2(3/4)^5=0{,}75640869140625.$$

**Ответ с округлением до трёх знаков — 0,756.**

```python
from math import comb

n = 7
p = 1 / 4
probability = sum(comb(n, k) * p**k * (1 - p)**(n - k) for k in range(3))
print(f"Вероятность выбрать не более двух устройств с подключением: {probability:.3f}")
```

**Задача 2** (30 баллов)

Дан закон распределения количества публикаций в которой группе телеграм-каналов:

$$
\begin{array}{|c|c|c|c|c|c|c|}
\hline X \text { (число публикаций) } & 0 & 1 & 2 & 3 & 4 & 5 \\
\hline \mathrm{P} \text { (вероятность) } & ? & 0,3 & 0,3 & 0,2 & 0,1 & 0,1 \\
\hline
\end{array}
$$

Найдите пропущенную вероятность для X = 0, математическое ожидание и дисперсию. Решите задачу "вручную" и с помощью python.

**Ручной расчёт**

Сумма вероятностей всех возможных значений равна единице, поэтому

$$P(X=0)=1-(0{,}3+0{,}3+0{,}2+0{,}1+0{,}1)=0.$$

Для математического ожидания умножаем каждое значение на его вероятность и складываем произведения.

$$E(X)=0\cdot 0+1\cdot 0{,}3+2\cdot 0{,}3+3\cdot 0{,}2+4\cdot 0{,}1+5\cdot 0{,}1=2{,}4.$$

Для дисперсии сначала найдём среднее значение квадрата числа публикаций.

$$E(X^2)=0^2\cdot 0+1^2\cdot 0{,}3+2^2\cdot 0{,}3+3^2\cdot 0{,}2+4^2\cdot 0{,}1+5^2\cdot 0{,}1=7{,}4.$$

$$D(X)=E(X^2)-(E(X))^2=7{,}4-2{,}4^2=1{,}64.$$

Ниже проверим дисперсию в Python по определению — как среднее квадратов отклонений от $E(X)$.

```python
from math import fsum

known = [0.3, 0.3, 0.2, 0.1, 0.1]
p_zero = 1 - fsum(known)
probabilities = [p_zero] + known

mean = sum(x * p for x, p in enumerate(probabilities))
variance = sum((x - mean)**2 * p for x, p in enumerate(probabilities))

print(f"P(X = 0) = {p_zero:g}")
print(f"E(X) = {mean:.1f}")
print(f"D(X) = {variance:.2f}")
```

**Задача 3** (25 баллов)


Компания приобрела 10 000 новых ноутбуков. Вероятность поломки одного ноутбука в течение месяца равна 0,0002. Предполагается, что поломки отдельных ноутбуков происходят независимо друг от друга.

**Вопрос 1**. Найдите вероятность того, что за месяц сломается ровно 1 ноутбук.
Решение объясните. (5 баллов)


**1. Ровно одна поломка.** Обозначим число сломавшихся за месяц ноутбуков через $X$. У каждого из 10 000 ноутбуков одинаковая вероятность поломки, а поломки независимы. Поэтому $X$ имеет биномиальное распределение с параметрами $n=10000$ и $p=0{,}0002$.

Сломаться может любой из 10 000 ноутбуков, а остальные 9 999 должны проработать весь месяц. Отсюда

$$P(X=1)=10000\cdot 0{,}0002\cdot (0{,}9998)^{9999}\approx 0{,}270671.$$

Ответ — примерно **27,07 %**.

**Вопрос 2**. Компания планирует ввести программу расширенной гарантии, по которой она будет бесплатно ремонтировать или заменять ноутбуки, сломавшиеся в течение месяца. Если общее количество поломок за месяц превысит 5 ноутбуков, компания понесет значительные дополнительные расходы.

а) Найдите вероятность того, что за месяц сломается более 5 ноутбуков.Решение объясните (10 баллов)

б) Если стоимость ремонта или замены одного ноутбука составляет 500 долларов, вычислите математическое ожидание общих расходов компании на поломки за один месяц. (10 баллов)

**2а. Вероятность значительных расходов.** Более пяти поломок означает шесть или больше. Противоположное событие включает все случаи от нуля до пяти поломок включительно.

$$P(X>5)=1- P(X\le 5)=1- \sum_{k=0}^{5} {10000 \choose k}(0{,}0002)^k(0{,}9998)^{10000-k}\approx 0{,}016553.$$

Вероятность превысить порог за месяц составляет примерно **1,66 %**.

**2б. Ожидаемые расходы.** В среднем за месяц ломается $E(X)=np=10000\cdot 0{,}0002=2$ ноутбука. Компания оплачивает каждый ремонт, поэтому общие расходы равны $C=500X$. По линейности математического ожидания $E(C)=500E(X)=500\cdot 2=1000$ долларов.

```python
from math import comb

n = 10_000
p = 0.0002
repair_cost = 500

def probability_of_k(k):
    return comb(n, k) * p**k * (1 - p)**(n - k)

more_than_five = 1 - sum(probability_of_k(k) for k in range(6))
print(f"Ровно одна поломка: {probability_of_k(1):.6%}")
print(f"Более пяти поломок: {more_than_five:.6%}")
print(f"Ожидаемые расходы за месяц: {repair_cost * n * p:.0f} долларов")
```
