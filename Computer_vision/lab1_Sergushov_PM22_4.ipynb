```python
image = array(Image.open('/content/GERMON.jpg').convert('L'))
```

```python
image
```

```python
im1, bin_edges = np.histogram(image, bins=256)
plt.plot(im1)
```

```python
def image_histogram_equalization(im, number_bins=256):
  imhist, bins = np.histogram(im.flatten(), number_bins, density=True)
  cdf = imhist.cumsum()
  cdf = 255*cdf/cdf[-1]
  image_equalized = np.interp(im.flatten(), bins[:-1], cdf)
  return image_equalized.reshape(im.shape), cdf
```

```python
im2, cdf = image_histogram_equalization(image)
```

```python
im_2 = Image.fromarray(np.uint8(im2))
im_2
```

```python
im_2, bin_edges = np.histogram(im2, bins=256)
plt.plot(im_2)
```

```python
while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Преобразуем кадр в оттенки серого
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Обнаружение краев с помощью алгоритма Кэнни
    edges = cv2.Canny(gray, threshold1=100, threshold2=200)

    cv2_imshow(frame)  # Оригинальное видео
    cv2_imshow(edges)  # Обнаруженные края
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
```

## Задание 3

Реализуйте собственный пример с сопоставлением шаблонов.

```python
target = cv2.imread('/content/GERMON.jpg', cv2.IMREAD_GRAYSCALE)
template = cv2.imread('/content/GERMON2.jpg', cv2.IMREAD_GRAYSCALE)
```

```python
cv2_imshow(target)
cv2_imshow(template)
```

```python
w, h = template.shape[::-1]

methods = [
    ('cv2.TM_CCOEFF', cv2.TM_CCOEFF),
    ('cv2.TM_CCOEFF_NORMED', cv2.TM_CCOEFF_NORMED),
    ('cv2.TM_CCORR', cv2.TM_CCORR),
    ('cv2.TM_CCORR_NORMED', cv2.TM_CCORR_NORMED),
    ('cv2.TM_SQDIFF', cv2.TM_SQDIFF),
    ('cv2.TM_SQDIFF_NORMED', cv2.TM_SQDIFF_NORMED)
]

for method_name, method in methods:
    img = target.copy()
    result = cv2.matchTemplate(img, template, method)
    min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(result)

    if method in [cv2.TM_SQDIFF, cv2.TM_SQDIFF_NORMED]:
        top_left = min_loc
    else:
        top_left = max_loc

    bottom_right = (top_left[0] + w, top_left[1] + h)
    cv2.rectangle(img, top_left, bottom_right, 255, 2)

    print(f"Метод: {method_name}")
    cv2_imshow(img)
    cv2.waitKey(0)

cv2.destroyAllWindows()
```

## Задание 6
Подберите произвольную цветную картинку для кластеризации и произведите кластеризацию цветов для значений количества кластеров 3, 5, 7, 9.

```python
image = cv2.imread('/content/cat.jpg')
cv2_imshow(image)
```

```python
img = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
Z = img.reshape((-1, 3))
Z = np.float32(Z)

criteria = (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER, 10, 1.0)

cluster_counts = [3, 5, 7, 9]
for K in cluster_counts:
    _, y_pred, centers = cv2.kmeans(Z, K, None, criteria, 10, cv2.KMEANS_RANDOM_CENTERS)
    centers = np.uint8(centers)
    res = centers[y_pred.flatten()]
    res2 = res.reshape((img.shape))

    plt.figure()
    plt.title(f'Кластеров: {K}')
    plt.imshow(res2)
    plt.axis('off')
    plt.show()
```
